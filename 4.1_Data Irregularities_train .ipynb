```python
import pandas as pd

#  Data with Outliers

df = pd.DataFrame({'price': [10, 180, 200, 210, 190, 205, 195, 220, 1000]})

# q1 and q3

q1 = df['price'].quantile(0.25)
q1
q3 = df['price'].quantile(0.75)
q3

# IQR

iqr = q3 - q1

# Lower and Upper Bound

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# fILTER

outliers_iqr = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]
clean_df_iqr = df[(df['price'] >= lower_bound) & (df['price'] <= upper_bound)]

print("IQR:")
print(outliers_iqr)
```

```python
import pandas as pd

df = pd.DataFrame({'price': [10, 180, 200, 210, 190, 205, 195, 220, 1000]})

# Z-Score 

mean = df['price'].mean()
std = df['price'].std()

df['z_score'] = (df['price'] - mean) / std


threshold = 2
outliers_zscore = df[df['z_score'].abs() > threshold]

print("Z-Score:")
print(outliers_zscore)
```
