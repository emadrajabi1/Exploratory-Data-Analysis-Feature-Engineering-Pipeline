Exercise 6: Identifying Structured vs. Semi-Structured Data
Why is raw_record not a simple "structured" table (like an Excel spreadsheet)?

How can you transform this record into a Pandas DataFrame where each purchase gets its own row? (Hint: Flatten user and expand purchases using pd.json_normalize with record_path and meta parameters).

```python
raw_record = {
    "id": 501,
    "user": {"name": "Sara", "country": "Iran"},
    "purchases": [{"item": "Laptop", "price": 1200}, {"item": "Mouse", "price": 25}]
}
```

```python
df6 = pd.json_normalize(
    raw_record,
    record_path = ['purchases'],
    meta= ['id',['user','name'],['user','country']]
)
df6
```

```python
df6 = df6.rename(columns={
    'user.name' : 'Username',
    'user.country' : 'Country'
})
df6
```

Exercise 7 (Comprehensive): Cleaning and Analyzing a Messy DataFrame
Use .info() and .dtypes to identify all data type issues in df7.

Convert salary to numeric (Note: "NaN" here is a string, not a real NaN — use pd.to_numeric(df7["salary"], errors="coerce")).

Convert join_date to datetime.

Convert is_manager to boolean (bool).

Convert gender to "category".

Finally, run .info() again to confirm all data types are corrected.

```python
df7 = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "gender": ["M", "F", "F", "M", "F"],
    "salary": ["5000", "6200", "NaN", "7100", "5900"],
    "join_date": ["2021-01-10", "2021-05-22", "2022-01-01", None, "2023-07-19"],
    "is_manager": [1, 0, 0, 1, 0]
})
df7
```

```python
df7.info()
```

```python
df7['salary'] = pd.to_numeric(df7['salary'], errors='coerce')
df7['join_date'] = pd.to_datetime(df7['join_date'])
df7['is_manager'] = df7['is_manager'].astype(bool)
df7['gender'] = df7['gender'].astype('category')
```

```python
df7.info()
```
