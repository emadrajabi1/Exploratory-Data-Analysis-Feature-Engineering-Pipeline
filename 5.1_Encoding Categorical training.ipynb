```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Ordinal Categories

df = pd.DataFrame({'Education': ['High School', 'Bachelor', 'Master', 'PhD']})


categories_order = [['High School', 'Bachelor', 'Master', 'PhD']]

encoder = OrdinalEncoder(categories=categories_order)
df['Education_Encoded'] = encoder.fit_transform(df[['Education']])

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green', 'Red']})

# One-Hot Categories

df_encoded = pd.get_dummies(df, columns=['Color'], drop_first=True)

print(df_encoded)
```

```python
import pandas as pd

df = pd.DataFrame({
    'City': ['London', 'Paris', 'London', 'Tokyo', 'Paris'],
    'Sales': [100, 90, 110, 120, 95]

})

# Target Encoding

mean_encoding = df.groupby('City')['Sales'].mean().to_dict()

df['City_Target_Encoded'] = df['City'].map(mean_encoding)

print(mean_encoding)
print(df)
```
